<h1><font color="#1a73e8">What Is Machine Learning?</font></h1>
<p><font size="3" color="#3c4043"><i>"Machine learning is about extracting knowledge from data."</i></font> <font size="2" color="#80868b">(Müller & Guido, p. 1)</font></p>
<p><font size="3" color="#3c4043">You already use it every day: film and music recommendations, spam filters and phone cameras that recognise faces all rely on machine learning.</font></p>

<h3><font color="#1a73e8">Writing rules by hand</font></h3>
<p><font size="3" color="#3c4043">Before machine learning, "intelligent" programs were built from hand-written <code>if</code> and <code>else</code> rules (p. 1). For our task, a rule-based program might say:</font></p>
<table width="100%" cellspacing="0" cellpadding="6">
<tr><td bgcolor="#5f6368" width="8" rowspan="3"></td><td bgcolor="#f1f3f4">&nbsp;</td></tr>
<tr><td bgcolor="#f1f3f4"><font size="2" color="#202124">if a review contains "love" or "delicious" → <b>positive</b></font></td></tr>
<tr><td bgcolor="#f1f3f4"><font size="2" color="#202124">if a review contains "hate" or "cold" → <b>negative</b></font></td></tr>
</table>
<p></p>
<table width="100%" cellspacing="0" cellpadding="6">
<tr><td bgcolor="#d93025" width="8" rowspan="7"></td><td bgcolor="#fce8e6">&nbsp;</td></tr>
<tr><td bgcolor="#fce8e6"><font size="3" color="#a50e0e"><b>Why hand-written rules fail</b></font></td></tr>
<tr><td bgcolor="#fce8e6"><font size="2" color="#202124"><b>1. They only fit one task.</b> Change the task slightly, and you may rewrite everything.</font></td></tr>
<tr><td bgcolor="#fce8e6"><font size="2" color="#202124"><b>2. You must already know how to decide.</b> Someone has to spell out every rule.</font></td></tr>
<tr><td bgcolor="#fce8e6"><font size="2" color="#80868b"><i>(Müller & Guido, p. 2)</i></font></td></tr>
<tr><td bgcolor="#fce8e6"><font size="2" color="#202124"><i>"I didn't love it."</i> contains "love", but it is <b>not</b> positive.</font></td></tr>
<tr><td bgcolor="#fce8e6"><font size="2" color="#202124"><i>"Friendly staff, would return."</i> matches <b>none</b> of the rules.</font></td></tr>
</table>

<h3><font color="#1a73e8">Learning from examples instead</font></h3>
<p><font size="3" color="#3c4043">Machine learning takes a different approach. Instead of writing the rules, we give the computer many reviews <b>together with the correct answer</b> for each one, and it works out the rules itself.</font></p>
<p><font size="3" color="#3c4043">Müller and Guido call this <b>supervised learning</b>: the algorithm learns from pairs of inputs and correct outputs, and can then produce an output</font> <font size="3" color="#3c4043"><i>"for an input it has never seen before without any help from a human"</i></font> <font size="2" color="#80868b">(p. 2)</font><font size="3" color="#3c4043">. It is called <i>supervised</i> because the correct answers act like a teacher.</font></p>

<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">Think of the question bank as a spreadsheet with two columns.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">One column holds the questions; the other holds their correct answers.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">First, the instructor copies each column into its own <b>separate list</b>:</font></td></tr>

<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">Then the instructor randomly picks <b>20%</b> of the questions, with their answers, for the exam.</font></td></tr>

<h3><font color="#1a73e8">The concept</font></h3>
<p><font size="3" color="#3c4043">Just like the instructor hides 20% of the questions to save them for the exam, we set aside 20% of our labelled reviews to save them for testing the model in Step 6. The model will train on the other 80% in Step 5. For example, if our file has 1,000 reviews, about 800 go to training and about 200 are kept aside for testing.</font></p>
<p><font size="3" color="#3c4043">Why? Imagine the model learned from the review <i>"The food was delicious and the staff were friendly."</i> If we then tested it on that same review, getting it right would tell us very little. It has already seen it. We want to know whether it works on reviews it has <b>never seen</b>.</font></p>

<h1><font color="#1a73e8">Step 4: Turn Text into Numbers</font></h1>
<table width="100%" cellspacing="0" cellpadding="6">
<tr><td bgcolor="#8430ce" width="8" rowspan="12"></td><td bgcolor="#f3e8fd">&nbsp;</td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#681da8" size="4"><b>Analogy – Step 4: Translating the questions</b></font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">Imagine a student who reads only their <b>native language</b>, such as Spanish.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">But the questions are written in another language, such as English.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">So the instructor must <b>translate</b> every question into the student's native language.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">For example, <i>"Which organ pumps blood around the body?"</i></font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">becomes <i>"¿Qué órgano bombea la sangre por el cuerpo?"</i></font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">The instructor builds a <b>dictionary</b> from the study questions and uses it to translate them.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">For example, the dictionary says <i>organ → órgano</i> and <i>blood → sangre</i>.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">Later, the exam questions are translated with <b>the same dictionary</b>.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">That way, the student sees them in the same form they studied.</font></td></tr>
<tr><td bgcolor="#f3e8fd"><font color="#202124" size="2">Without translation, the student could neither study nor answer the exam.</font></td></tr>
</table>
<h3><font color="#1a73e8">The concept</font></h3>
<p><font size="3" color="#3c4043">Just like the instructor translates the questions into the student's native language, we translate every review into <b>numbers</b>, the native language of computers. A machine learning model can only calculate with numbers, and calculating is how it works: in <b>Step 5</b> it learns by adjusting numbers, and in <b>Step 6</b> it predicts by adding numbers up. For example, the review <i>"The food was delicious"</i> will become a row of numbers like <code>[0 1 1 0 0 0 1 1]</code>. You will see how in Cell 10.</font></p>
<p><font size="3" color="#3c4043">The test reviews must be translated <b>the same way</b> as the training reviews, so the model sees them in the same form it learned from.</font></p>
<p><font size="3" color="#3c4043">The simplest way to translate text into numbers is called <b>bag of words</b>: it ignores word order and counts how often each word appears in each review. We try it on three tiny reviews first. It has three steps:</font></p>
<table width="100%" cellspacing="0" cellpadding="6">
<tr><td bgcolor="#5f6368" width="8" rowspan="4"></td><td bgcolor="#f1f3f4">&nbsp;</td></tr>
<tr><td bgcolor="#f1f3f4"><font size="2" color="#202124"><b>1. Split:</b> break each review into words, called <b>tokens</b>.</font></td></tr>
<tr><td bgcolor="#f1f3f4"><font size="2" color="#202124"><b>2. Build a vocabulary:</b> list every word that appears in any review.</font></td></tr>
<tr><td bgcolor="#f1f3f4"><font size="2" color="#202124"><b>3. Count:</b> for each review, count how often each vocabulary word appears.</font></td></tr>
</table>
<p></p>
<table width="100%" cellspacing="0" cellpadding="6">
<tr><td bgcolor="#0d652d" width="8" rowspan="3"></td><td bgcolor="#188038">&nbsp;</td></tr>
<tr><td bgcolor="#188038"><font size="4" color="#ffffff"><b>YOUR TURN – Cell 10: Bag of words on three tiny reviews</b></font></td></tr>
<tr><td bgcolor="#188038"><font size="2" color="#ffffff">1. Copy the code below → 2. Paste it into the empty code cell underneath → 3. Click ▶ to run it</font></td></tr>
</table>

```python
# YOUR TURN – Cell 10: copy → paste into the empty cell below → click ▶ to run
from sklearn.feature_extraction.text import CountVectorizer

tiny_reviews = ["The food was delicious",
                "The soup was cold",
                "Delicious soup, friendly staff"]

counter = CountVectorizer()
counter.fit(tiny_reviews)                              # steps 1 and 2: split, build vocabulary
print("Vocabulary:", counter.get_feature_names_out())
print(counter.transform(tiny_reviews).toarray())       # step 3: count
```

<h3><font color="#1a73e8">TF‑IDF: giving informative words more weight</font></h3>
<p><font size="3" color="#3c4043">Counting treats every word the same, but words like "the" and "was" appear in almost every review and say nothing about sentiment. <b>TF‑IDF</b> fixes this: it gives more weight to words that appear in only a few reviews. In our tiny reviews, <i>friendly</i> and <i>staff</i> (in one review only) would get a higher weight than <i>delicious</i> and <i>soup</i> (in two reviews).</font></p>
<p></p>
<table width="100%" cellspacing="0" cellpadding="6">
<tr><td bgcolor="#8430ce" width="8" rowspan="1"></td><td bgcolor="#f3e8fd"><font color="#202124" size="2"><b>Exam analogy:</b> the tiny example was like practice. Now the instructor translates every real question.</font></td></tr>
</table>
<p><font size="3" color="#3c4043">Now we use TF‑IDF on our real reviews. One rule matters: the vocabulary is learned from the <b>training reviews only</b>. The test reviews stay unseen, so we only convert them, using the vocabulary already learned.</font></p>
<table width="100%" cellspacing="0" cellpadding="6">
